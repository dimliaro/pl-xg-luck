```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# ── Ερωτήσεις ────────────────────────────────────────────────────────────────
questions = [
    {
        "topic": "Topic 2 — Duplicate Records",
        "question": "Βρες ονόματα υπαλλήλων που εμφανίζονται παραπάνω από μια φορά.",
        "answer": """SELECT name, COUNT(*) AS count
FROM employees
GROUP BY name
HAVING COUNT(*) > 1"""
    },
    {
        "topic": "Topic 4 — Unmatched Records",
        "question": "Βρες πελάτες που δεν έχουν κάνει καμία παραγγελία.",
        "answer": """SELECT c.customer_id, c.name
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL"""
    },
    {
        "topic": "Topic 1 — Second Highest Salary",
        "question": "Βρες τον 2ο υψηλότερο μισθό. Αν δεν υπάρχει, επέστρεψε NULL.",
        "answer": """SELECT (
    SELECT DISTINCT salary
    FROM employees
    ORDER BY salary DESC
    LIMIT 1 OFFSET 1
) AS SecondHighestSalary"""
    },
    {
        "topic": "Topic 3 — Top N per Group",
        "question": "Βρες τον υπάλληλο με τον υψηλότερο μισθό σε κάθε department.",
        "answer": """SELECT name, department, salary
FROM (
    SELECT name, department, salary,
           RANK() OVER (PARTITION BY department
                        ORDER BY salary DESC) AS rnk
    FROM employees
) sub
WHERE rnk = 1"""
    },
    {
        "topic": "Topic 5 — Month-over-Month",
        "question": "Υπολόγισε το revenue κάθε μήνα και τη διαφορά από τον προηγούμενο.",
        "answer": """SELECT month, revenue,
       LAG(revenue, 1) OVER (ORDER BY month) AS prev_revenue,
       revenue - LAG(revenue, 1) OVER (ORDER BY month) AS change
FROM sales"""
    },
]

# ── State ─────────────────────────────────────────────────────────────────────
state = {"idx": 0, "score": 0, "answered": False}

# ── UI Elements ───────────────────────────────────────────────────────────────
topic_label    = widgets.HTML()
question_label = widgets.HTML()
answer_box     = widgets.Textarea(placeholder="Γράψε το query σου εδώ...",
                                  layout=widgets.Layout(width="100%", height="150px"))
submit_btn     = widgets.Button(description="Υποβολή", button_style="primary")
show_btn       = widgets.Button(description="Δες Λύση", button_style="warning")
next_btn       = widgets.Button(description="Επόμενο →", button_style="success", disabled=True)
feedback_out   = widgets.Output()
score_label    = widgets.HTML()

def load_question():
    idx = state["idx"]
    q   = questions[idx]
    state["answered"] = False
    answer_box.value  = ""
    next_btn.disabled = True
    topic_label.value    = f"<b style='color:#5340c9'>{q['topic']}</b> &nbsp;|&nbsp; Ερώτηση {idx+1}/{len(questions)}"
    question_label.value = f"<div style='font-size:14px;padding:10px;background:#f8f8fc;border-radius:6px;border-left:4px solid #5340c9'>{q['question']}</div>"
    score_label.value    = f"<b>Score: {state['score']}/{len(questions)}</b>"
    with feedback_out:
        clear_output()

def on_submit(b):
    if state["answered"]:
        return
    state["answered"] = True
    next_btn.disabled = False
    
    user_query = answer_box.value.strip()
    correct_query = questions[state["idx"]]["answer"].strip()
    
    try:
        user_result    = pd.read_sql(user_query, conn).reset_index(drop=True)
        correct_result = pd.read_sql(correct_query, conn).reset_index(drop=True)
        
        # Συγκρίνουμε αποτελέσματα — αγνοούμε ονόματα στηλών και σειρά γραμμών
        user_sorted    = user_result.sort_values(by=user_result.columns.tolist()).reset_index(drop=True)
        correct_sorted = correct_result.sort_values(by=correct_result.columns.tolist()).reset_index(drop=True)
        
        user_vals    = user_sorted.values.tolist()
        correct_vals = correct_sorted.values.tolist()
        
        is_correct = user_vals == correct_vals
        
        if is_correct:
            state["score"] += 1
        
        with feedback_out:
            clear_output()
            if is_correct:
                print("✅ Σωστό!")
                display(user_result)
            else:
                print("❌ Όχι ακριβώς — τα αποτελέσματα διαφέρουν.")
                print("\nΤο δικό σου:")
                display(user_result)
                print("\nΑναμενόμενο:")
                display(correct_result)
                
    except Exception as e:
        with feedback_out:
            clear_output()
            print(f"❌ Σφάλμα στο query: {e}")
        next_btn.disabled = False
    
    score_label.value = f"<b>Score: {state['score']}/{len(questions)}</b>"

submit_btn.on_click(on_submit)

def on_show(b):
    with feedback_out:
        clear_output()
        print("📋 Λύση:")
        print(questions[state["idx"]]["answer"])
    next_btn.disabled = False

def on_next(b):
    state["idx"] += 1
    if state["idx"] >= len(questions):
        with feedback_out:
            clear_output()
            print(f"🏁 Τέλος! Score: {state['score']}/{len(questions)}")
        topic_label.value    = "<b>Τέλος quiz!</b>"
        question_label.value = ""
        answer_box.disabled  = True
        submit_btn.disabled  = True
        show_btn.disabled    = True
        next_btn.disabled    = True
    else:
        load_question()

submit_btn.on_click(on_submit)
show_btn.on_click(on_show)
next_btn.on_click(on_next)

# ── Layout ────────────────────────────────────────────────────────────────────
load_question()
display(widgets.VBox([
    score_label,
    topic_label,
    question_label,
    answer_box,
    widgets.HBox([submit_btn, show_btn, next_btn]),
    feedback_out,
]))
```
