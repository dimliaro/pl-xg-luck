```python
from understatapi import UnderstatClient

with UnderstatClient() as understat:
    teams = understat.league(league='EPL').get_team_data(season = "2026")


```

```python
import pandas as pd

rows = []

for team_id, team_info in teams.items():
    team_name = team_info['title']

    for matches in team_info['history']:
        row = {"team": team_name, **matches}
        rows.append(row)

df = pd.DataFrame(rows)
print(df.head(10))
print(df.shape)

```

```python
luck = df.groupby('team')[['pts','xpts']].sum()

luck['luck'] = luck['pts'] - luck['xpts']

luck.sort_values('luck', ascending = False)
print(luck)
```
